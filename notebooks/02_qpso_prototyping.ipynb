```python
import numpy as np
import matplotlib.pyplot as plt
from backend.app.optimizers.qpso import qpso_solver
from backend.app.graph.graph_builder import graph_builder

# 1. Initialize toy VRP
nodes, dist_mat, time_mat, G = graph_builder.generate_synthetic_graph(node_count=15)
route, stats = qpso_solver.solve_single_tour(
    nodes=nodes, dist_matrix=dist_mat, time_matrix=time_mat, start_hour=9.0, vehicle_capacity=20
)

print(f"QPSO Tour Optimality: Best Energy = {stats['best_energy']}, Tunnels = {stats['tunnels']}")
```

```python
# 2. Plot Convergence History and Beta Decay
fig, ax1 = plt.subplots(figsize=(10, 4))

ax1.plot(stats["history"], color='#00f3ff', linewidth=2, label='QPSO Energy (Cost)')
ax1.set_xlabel('Iteration')
ax1.set_ylabel('Objective Function Cost', color='#00f3ff')
ax1.tick_params(axis='y', labelcolor='#00f3ff')
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(stats["beta_history"], color='#ff9100', linestyle='--', label=r'Contraction-Expansion $\beta(t)$')
ax2.set_ylabel(r'$\beta(t)$ Coefficient', color='#ff9100')
ax2.tick_params(axis='y', labelcolor='#ff9100')

plt.title('QPSO Convergence & Quantum Parameter Schedule $\\beta(t)$')
plt.show()
```
