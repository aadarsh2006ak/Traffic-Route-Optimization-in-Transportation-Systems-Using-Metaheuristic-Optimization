```python
# 2. Plot 24-Hour Congestion Profile
profile = traffic_simulator.get_24h_profile()
hours = [p["hour"] for p in profile]
multipliers = [p["congestion_multiplier"] for p in profile]

plt.figure(figsize=(10, 4))
plt.plot(hours, multipliers, color='#00e5ff', linewidth=2.5, label=r'Congestion Factor $\theta(t)$')
plt.axvline(8.5, color='#ff2b2b', linestyle='--', alpha=0.7, label='Morning Rush (8:30 AM)')
plt.axvline(18.0, color='#ff9100', linestyle='--', alpha=0.7, label='Evening Rush (6:00 PM)')
plt.title('24-Hour Urban Traffic Congestion Profile $\\theta(t)$', fontsize=12)
plt.xlabel('Hour of Day')
plt.ylabel('Multiplier $\\theta(t)$')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
