This notebook performs statistical comparison sweeps:
- **QPSO (Ours)** vs **Classical PSO** vs **Genetic Algorithm** vs **Ant Colony** vs **Exact Solver**.
- Generates convergence speed plots, runtime boxplots, and scalability analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
from backend.app.services.benchmark_service import benchmark_service
from backend.app.graph.graph_builder import graph_builder

# 1. Run 20-Node Multi-Algorithm Benchmark
nodes, dist_mat, time_mat, G = graph_builder.generate_synthetic_graph(node_count=20)
res = benchmark_service.run_benchmark_suite(
    start_node=nodes[0],
    stops_data=nodes[1:],
    algorithms_to_run=["QPSO", "Classical PSO", "Genetic Algorithm", "Ant Colony", "Exact Solver"],
    trials_per_algo=3
)

df_summary = pd.DataFrame(res["summary_table"])
df_summary[["algorithm", "total_cost", "distance_km", "duration_min", "runtime_sec", "stability_std"]]
```

```python
# 2. Plot Overlay Convergence Traces
plt.figure(figsize=(10, 5))
colors = {"QPSO": "#00f3ff", "Classical PSO": "#ff0055", "Genetic Algorithm": "#a855f7", "Ant Colony": "#10b981", "Exact Solver": "#f59e0b"}

for algo, trace in res["convergence"].items():
    plt.plot(trace, label=algo, color=colors.get(algo, "#9ca3af"), linewidth=2)

plt.title("Convergence Trace Comparison (Best Objective Energy vs Iteration)")
plt.xlabel("Iteration")
plt.ylabel("Route Cost (Distance + Time + Penalty)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
